```python
import numpy as np


# Activation function (bipolar step)
def activation(x):
    return np.where(x >= 0, 1, -1)


# MADALINE class
class Madaline:
    def __init__(self, input_size, hidden_neurons, lr=0.1):
        self.lr = lr

        # Weights initialization
        self.W = np.random.randn(hidden_neurons, input_size)
        self.b = np.random.randn(hidden_neurons)

        self.V = np.random.randn(hidden_neurons)
        self.bo = np.random.randn(1)


    def forward(self, x):
        # Hidden layer (ADALINE units)
        z_in = np.dot(self.W, x) + self.b
        z = activation(z_in)

        # Output layer
        y_in = np.dot(self.V, z) + self.bo
        y = activation(y_in)

        return z, y


    def train(self, X, y, epochs=20):
        for epoch in range(epochs):
            total_error = 0

            for i in range(len(X)):
                x = X[i]
                target = y[i]

                z, output = self.forward(x)

                if output != target:
                    total_error += 1

                    # MADALINE Rule (adjust hidden neurons)
                    for j in range(len(self.W)):
                        zin = np.dot(self.W[j], x) + self.b[j]

                        # Update only misclassified neurons
                        if activation(zin) != target:
                            self.W[j] += self.lr * target * x
                            self.b[j] += self.lr * target

            print(f"Epoch {epoch+1}, Error: {total_error}")


    def predict(self, X):
        results = []
        for x in X:
            _, y = self.forward(x)
            results.append(y)
        return np.array(results)


# =========================
# DATASET (XOR Problem)
# =========================

# Bipolar inputs
X = np.array([
    [-1, -1],
    [-1,  1],
    [ 1, -1],
    [ 1,  1]
])

# XOR output (bipolar)
y = np.array([-1, 1, 1, -1])


# =========================
# TRAIN MODEL
# =========================

model = Madaline(input_size=2, hidden_neurons=2, lr=0.1)
model.train(X, y, epochs=20)


# =========================
# TEST MODEL
# =========================

predictions = model.predict(X)

print("\nPredictions:")
for i in range(len(X)):
    print(f"Input: {X[i]} -> Predicted: {predictions[i]}, Actual: {y[i]}")
```
